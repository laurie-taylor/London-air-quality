## Original Location Mapping for Data Collection:

- I originally had 3 locations: Marylebone Road (high recorded pollution), _Bloomsbury (mid-range pollution)_ and Richmond (low pollution). I hoped that this way I could compare weekday and weekend pollution patterns between constrasting urban environments. 
- However, during Data Transformation sanity checks, I noticed that Marylebone and Bloomsbury returned identical 5-year datasets across all pollutants (see **NB02 Step 6 Personal Reflection** for a summary of how I tested this). This suggested that both pairs of coordinates were mapped to the same OpenWeather model grid cell. Therefore, it wasn't a coding error, but a structural property of the dataset. This emphasised to me the importance of validating spatial independence between locations before conducting Data Transformation.
- To investigate further, I tried a few other locations in Bloomsbury's place (South Ken -> same results as Marylebone; Lewisham and Elephant & Castle -> identical datasets). This told me that several London coordinates had shared cells, in both central and outer-London. 
- As a result, I omitted the 2nd location and focussed on Marylebone Rd and Richmond. This improved the methodological robustness of my project, without taking too much time.
- By evaluating the data source, I learned that unique coordinates don't guarantee distinct datasets when working with gridded environmental APIs. Independence must be confirmed by comapring returned time series.
